```python
# Standardize column names to lowercase snake_case

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r'[^a-zA-Z0-9]+', '_', regex=True)
    .str.strip('_')
)

# Check updated column names
df.columns.tolist()
```

```python
df.info()
```

```python
# Convert date columns to datetime
date_cols = ['startdate', 'exitdate', 'dob']

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# Convert categorical columns
category_cols = [
    'businessunit', 'employeestatus', 'employeetype',
    'payzone', 'employeeclassificationtype',
    'terminationtype', 'departmenttype', 'division',
    'state', 'gendercode', 'racedesc', 'maritaldesc',
    'performance_score'
]

for col in category_cols:
    df[col] = df[col].astype('category')

# Convert location code to string
df['locationcode'] = df['locationcode'].astype('string')

# Convert text columns to string
text_cols = [
    'firstname', 'lastname', 'title', 'supervisor',
    'ademail', 'terminationdescription',
    'jobfunctiondescription'
]

for col in text_cols:
    df[col] = df[col].astype('string')

# Check updated data types
df.info()
```

```python
# Check future StartDate values
df[df['startdate'] > pd.Timestamp.today()]

# Check future ExitDate values
df[df['exitdate'] > pd.Timestamp.today()]

# Check ExitDate earlier than StartDate
df[df['exitdate'] < df['startdate']]

# Check employees whose DOB is after StartDate
df[df['dob'] > df['startdate']]
```

```python
# Calculate employee age
today = pd.Timestamp.today()

df['age'] = ((today - df['dob']).dt.days / 365.25).astype('float').round(0)

# Identify invalid ages
df[(df['age'] < 18) | (df['age'] > 100)]
```

```python
# Check invalid or duplicate employee IDs
df[df['empid'].isna()]

df[df['empid'].duplicated(keep=False)]
```

```python
# Check rating distribution
df['current_employee_rating'].value_counts()

# Identify ratings outside expected range (1–5)
df[
    ~df['current_employee_rating'].between(1, 5)
]
```

```python
# Check available performance categories
df['performance_score'].value_counts(dropna=False)
```

```python
# Check valid employee statuses
df['employeestatus'].value_counts(dropna=False)
```

```python
# Identify invalid email formats
df[
    ~df['ademail'].str.match(
        r'^[\w\.-]+@[\w\.-]+\.\w+$',
        na=False
    )
]
```

```python
# Standardize email addresses
df['ademail'] = (
    df['ademail']
    .str.strip()
    .str.lower()
    .str.replace(' ', '', regex=False)
)
```

```python
# Identify remaining invalid email addresses
invalid_emails = df[
    ~df['ademail'].str.match(
        r'^[\w\.-]+@[\w\.-]+\.\w+$',
        na=False
    )
]

invalid_emails[['empid', 'ademail']]
```

```python
# Check leading/trailing whitespace
for col in df.select_dtypes(include=['object', 'string', 'category']).columns:
    print(col, df[col].astype('string').str.contains(
        r'^\s|\s$', na=False
    ).sum())
```

```python
# Check missing values
print(df.isnull().sum())

# Check duplicate records
print("Duplicate rows:", df.duplicated().sum())

# Check data types
df.info()
```

```python
df.head()
```

```python
# Convert age to numeric
df["age"] = pd.to_numeric(df["age"], errors="coerce")

# Check whether any ages contain decimal values
fractional_ages = df.loc[
    df["age"].notna() & (df["age"] % 1 != 0),
    "age"
]

print("Fractional age values:", fractional_ages.unique())

# Convert to nullable integer if all values are whole numbers
if fractional_ages.empty:
    df["age"] = df["age"].astype("Int64")

# Export cleaned CSV
df.to_csv(
    "employee_data_cleaned.csv",
    index=False,
    encoding="utf-8"
)
```
