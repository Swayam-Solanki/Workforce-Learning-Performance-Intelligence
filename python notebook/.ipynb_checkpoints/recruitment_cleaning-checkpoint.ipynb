```python
df = pd.read_csv('../data/raw/recruitment_data.csv')
```

```python
# Convert date columns
date_cols = ['Application Date', 'Date of Birth']

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# Convert categorical columns
category_cols = [
    'Gender', 'City', 'State', 'Country',
    'Education Level', 'Job Title', 'Status'
]

for col in category_cols:
    df[col] = df[col].astype('category')

# Convert text and identification columns
string_cols = [
    'First Name', 'Last Name', 'Phone Number',
    'Email', 'Address', 'Zip Code'
]

for col in string_cols:
    df[col] = df[col].astype('string')

# Check updated data types
df.info()
```

```python
# Check future application dates
df[df['Application Date'] > pd.Timestamp.today()]

```

```python
# Check future DOB values
df[df['Date of Birth'] > pd.Timestamp.today()]

# Check applicants below 18 or above 100 years
today = pd.Timestamp.today()

df['Age'] = ((today - df['Date of Birth']).dt.days / 365.25).round(0)

df[(df['Age'] < 18) | (df['Age'] > 100)]
```

```python
# Identify negative or unrealistic experience
df[
    (df['Years of Experience'] < 0) |
    (df['Years of Experience'] > 60)
]
```

```python
# Identify zero or negative salaries
df[df['Desired Salary'] <= 0]
```

```python
df['Gender'].value_counts(dropna=False)
```

```python
df['Education Level'].value_counts(dropna=False)
```

```python
df.info()
```

```python
# Identify text columns
text_cols = [
    'First Name', 'Last Name', 'Gender',
    'Email', 'Address', 'City', 'State',
    'Country', 'Education Level', 'Job Title', 'Status'
]

# Remove unnecessary spaces
for col in text_cols:
    df[col] = df[col].str.strip()

# Standardize email addresses
df['Email'] = df['Email'].str.lower()

# Standardize capitalization
for col in [
    'First Name', 'Last Name', 'City',
    'State', 'Country', 'Education Level',
    'Job Title'
]:
    df[col] = df[col].str.title()

# Check updated values
for col in text_cols:
    print(f"\n{col}:")
    print(df[col].value_counts().head(20))
```
